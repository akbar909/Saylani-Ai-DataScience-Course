```python
list =["a", 1, 1, 40.5, True, "hello"]
print(list)
type(list)
```

```python
Fruits = ["apple", "banana", "cherry"]
print(Fruits[2])
type(Fruits[2])
```

```python
FruitBasket = ["apple", "banana","apple", "cherry","cherry"]
FruitBasket[-1] = "mango"
FruitBasket
FruitBasket.index("cherry")
```

```python
FruitBasket.insert(1, "lychee")
FruitBasket
```

```python
FruitBasket.append("Kiwi")
FruitBasket
```

```python
FruitBasket.extend(["peach", "Guava", "anaar"])
FruitBasket
```

```python
FruitBasket.pop(-1)
FruitBasket
```

```python
FruitBasket.remove("mango")
FruitBasket
```

```python
FruitBasket.sort()
FruitBasket
```

```python
FruitBasket.sort(reverse=True)
FruitBasket
```
