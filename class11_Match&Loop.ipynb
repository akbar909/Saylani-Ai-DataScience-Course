```python
day = "monday1"

match day:
    case "saturday":
        print("sat")
    case "sunday":
        print("sun")
    case "firday":
        print("fri")
    case "monday":
        print("mon")
    case "tuesday":
        print("tue")
    case "thursday":
        print("thu")
    case "wednesday":
        print("web")
    case _:
        print("not found")
```

```python
tableNum = int(input("Entar a num"))
steps = 1

while steps <=10:
    print(f"{tableNum} * {steps} = {tableNum*steps}")
    
    steps +=1
```
