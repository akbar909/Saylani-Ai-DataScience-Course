```python
while True:
    startingNum = int(input("Enter starting number: "))
    endingNum = int(input("Enter ending number: "))
    numelist = []
    sumList = 0
    if startingNum < endingNum:
        while startingNum <= endingNum:
            if startingNum % 2 == 1:
                numelist.append(startingNum)
                sumList += startingNum
            startingNum += 1
        print("Odd numbers:", numelist)
        print("Sum of odd numbers:", sumList)
        print("length of odd numbers:", len(numelist))
        print("Average of odd numbers:", sumList / len(numelist))
        break
    else:
        print("Starting number should be less than ending number")
```

```python
while True:
    startingNum = int(input("Enter starting number: "))
    endingNum = int(input("Enter ending number: "))
    choise = input("Enter choise EVEN or ODD: ")
    choise = choise.lower()
    numelist = []
    sumList = 0
    if startingNum < endingNum:
        if choise == "even":
          while startingNum <= endingNum:
            if startingNum % 2 == 0:
                numelist.append(startingNum)
                sumList += startingNum
            startingNum += 1
        elif choise == "odd":
          while startingNum <= endingNum:
            if startingNum % 2 == 1:
                numelist.append(startingNum)
                sumList += startingNum
            startingNum += 1
        print("Odd numbers:", numelist)
        print("Sum of odd numbers:", sumList)
        print("length of odd numbers:", len(numelist))
        print("Average of odd numbers:", sumList / len(numelist))
        break
    else:
        print("Starting number should be less than ending number")
```
