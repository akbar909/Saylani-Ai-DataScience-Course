For Loop

```python
fruits = ["Apple","Orange", "Banana", "Peach", "lychee"]
```

```python
for i in fruits:
    print(i)
    if i == "Banana":
      break
```

```python
for i in fruits:
    if i == "Banana":
      break
    print(i)
```

```python
for i in fruits:
    print(i)
    if i == "Banana":
      continue
```

```python
for i in fruits:
    if i == "Banana":
      continue
    print(i)
```

```python
words = ["Good", "Bad"]
```

```python
for i in fruits:
  for j in words:
    print(i, j)
```

```python
for i in words:
  for j in fruits:
    print(i,j)
```

```python
num1 = int(input("Enter num 1 : "))
num2 = int(input("Enter num 2 : "))

sum = 0
for i in range(num1, num2+1):
  if i%2 ==0:
    sum += i
print(f"The sum of even numbers from range {num1} to {num2} is {sum}")
```

```python
name = "MUHAMMAD DANIAL SIDDIQUI"
vowels = "aeiouAEIOU"
count = 0
for i in name:
  if  i in vowels:
    count +=1
print(f"There are {count} vowels in {name}")
```
