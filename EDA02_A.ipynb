```python
print(f"Shape of DataFrame before outlier removal: {df.shape}")

Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR
lower_bound = Q1 - 1.5 * IQR

df = df[(df['price'] >= lower_bound) & (df['price'] <= upper_bound)]

print(f"Shape of DataFrame after outlier removal: {df.shape}")
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df['price'] = scaler.fit_transform(df[['price']])

print("Price column normalized using Min-Max scaling.")
print(df['price'].head())
```

```python
import statsmodels.formula.api as smf
import statsmodels.api as sm

# Perform a multi-factor ANOVA
# C() is used to treat variables as categorical
# We start with main effects and can add interaction terms if needed.
model = smf.ols('price ~ C(view) + C(grade) + C(condition)', data=grouped_data).fit()
anova_table = sm.stats.anova_lm(model, typ=2)

print("ANOVA Table for View, Grade, and Condition on Price (from grouped_data):")
print(anova_table)
```
