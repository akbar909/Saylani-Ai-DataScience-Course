```python
def multiply(num1:int = 2, num2:int = 4):
    x = num1 * num2 
    return x
multiply(3)
```

```python
def addition(a):
    ad = a+a
    return ad
addition(2)
```

```python
def factorial(n):
    if n==0 or n==1:
        return 1
    else:
        return n * factorial(n-1)
factorial(5)
```

```python
def summation(n):
    if (n > 1):
        result = n + summation(n-1)
        return result
    else:
        return 0
summation(3)
```

```python
dic = {
   "Std1" : {"name" : "Akbar", "age": 20 },
   "Std2" : {"name" : "Danial", "age": 30 },
   "Std3" : {"name" : "GAkbar", "age": 31 },
   "Std4" : {"name" : "Ghulam_Akbar", "age": 28 }
}
```

```python
dic
```

```python
dic["Std1"]["age"]
```

```python
oldest_student_data = None
max_age = 0 
for student_data in dic.values():
    current_age = student_data["age"]
    if current_age > max_age:
        max_age = current_age
        oldest_student_data = student_data
print(f"The student with the maximum age is: {oldest_student_data['name']} (Age: {max_age})")
```
