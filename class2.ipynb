```python
celsius = float(input("Enter Celsius"))
fahrenheit = float(input("Enter hafrenheit"))
kelvin = float(input("Enter kelvin"))

celsiusTofahrenheit= (fahrenheit-32)/1.8
fahrenheitTocelsius= 1.8*celsius+32
kelvinTocelsius= kelvin-273.15
celsiusTokelvin= celsius+273.15


print(f"{celsius}C Celsius into fahrenheit {celsiusTofahrenheit}F")
print(f"{fahrenheit}F fahrenheit into Celsius {fahrenheitTocelsius}C")
print(f"{kelvin}K kelvin into Celsius {kelvinTocelsius}C")
print(f"{celsius}C Celsius into Kelvin {celsiusTokelvin}F")
```
