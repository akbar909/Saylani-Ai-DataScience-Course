```python
a = 7
b = 6
```

```python
if a < b: 
    print(f"{a} is less than {b}")
elif a > b:
    print(f"{a} is greater than {b}")
elif a == b: 
    print(f"{a} is equals to {b}")
else: 
    print("invalid")
    
    
```
