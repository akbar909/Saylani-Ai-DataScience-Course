```python
import random
```

```python
print("====== Welcome to Number Guessing Game ======")
while True:
    minNum = input("Enter Minimum Number : ")
    maxNum = input("Enter Maximum Number : ")
    if not minNum.isdigit() and not maxNum.isdigit():
        print("Enter valid input type, Numbers Only")
        continue
    else:
        minNum = int(minNum)
        maxNum = int(maxNum)
    
    
    if maxNum < minNum:
        print("Enter Maximum number greater than Minimum number")
    else:
        randNum = random.randint(minNum,maxNum)
        
        while True:
            difficultyLevel = input("Enter Difficulty Level (EASY OR MEDUIM OR HARD): ").lower()
            if difficultyLevel != "easy" and difficultyLevel != "meduim" and difficultyLevel != "hard":
                print("Enter valid input Type EASY // MEDUIM // HARD")
            else:
                if difficultyLevel == "easy":
                    chances = 20
                elif difficultyLevel == "meduim":
                    chances = 10
                elif difficultyLevel == "hard":
                    chances = 5
                break
        print(f"Apne '{difficultyLevel}' Difficulty Level select kiya hai isliye aapke paas {chances} chances hain")
        while True:
            hint = input("Hint Chahiye ? YES/NO : ").lower()
            if hint != "yes" and hint != "no":
                print("Enter YES or NO only")
            else:
                if hint =="yes":
                    if randNum%2 ==0:
                        print("Random Number is Even")
                    else:
                        print("Random Number is Odd")
                else:
                    print("Ok! no problem")
                break
        
        while chances > 0:
            
            print(f"{chances} chances left")
            userInput = input("Enter Num to guess : ")
            if not userInput.isdigit():
                print("Enter number only : ")
                continue
            else:
                userInput = int(userInput)
            if userInput > randNum:
                print("To High")
            elif userInput < randNum:
                print("Too Low")
            else:
                print("Correct! You Win")
                break
            chances -=1
            if chances == 0:
                print("You Lost")
        cont = input("Do you want to plat again ? YES/NO :").lower()
        if cont != "yes":
            break
                
```
