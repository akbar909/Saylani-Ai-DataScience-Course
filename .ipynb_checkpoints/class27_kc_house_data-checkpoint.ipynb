```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group by date and calculate the mean price
price_by_date = df.groupby('date')['price'].mean().reset_index()

plt.figure(figsize=(12, 6))
sns.lineplot(data=price_by_date, x='date', y='price')
plt.title('Average House Price Distribution by Date')
plt.xlabel('Date')
plt.ylabel('Average Price')
# plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='sqft_living', y='price')
plt.title('House Price vs. Square Footage of Living Area')
plt.xlabel('Square Footage of Living Area')
plt.ylabel('Price')
plt.grid(True)
plt.show()
```

```python
print(f"Shape of DataFrame before outlier removal: {df.shape}")

Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR
lower_bound = Q1 - 1.5 * IQR

df = df[(df['price'] >= lower_bound) & (df['price'] <= upper_bound)]

print(f"Shape of DataFrame after outlier removal: {df.shape}")
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df['price'] = scaler.fit_transform(df[['price']])

print("Price column normalized using Min-Max scaling.")
print(df['price'].head())
```

```python
raw_df = pd.read_csv('kc_house_data.csv')
original_prices = raw_df['price']
print("Original 'price' column loaded into 'original_prices'.")
print(original_prices.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15, 6))

# Histogram for original prices
plt.subplot(1, 2, 1)
sns.histplot(original_prices, kde=True, color='blue')
plt.title('Original Price Distribution')
plt.xlabel('Price')
plt.ylabel('Frequency')

# Histogram for processed prices
plt.subplot(1, 2, 2)
sns.histplot(df['price'], kde=True, color='green')
plt.title('Processed (Outlier Removed & Normalized) Price Distribution')
plt.xlabel('Normalized Price')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
df['bathrooms'] = df['bathrooms'].round().astype(int)
print("Converted 'bathrooms' column to integer type after rounding.")
print(df['bathrooms'].head())
```

```python
df['floors'] = df['floors'].round().astype(int)
print("Converted 'floors' column to integer type after rounding.")
print(df['floors'].head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

categorical_cols = ['bedrooms', 'bathrooms', 'floors', 'waterfront', 'view', 'condition', 'grade', 'zipcode']

for col in categorical_cols:
    print(f"\n--- Unique values and counts for '{col}' ---")
    print(df[col].value_counts().sort_index())

    plt.figure(figsize=(10, 6))
    sns.countplot(data=df, x=col, hue=col, palette='viridis', legend=False)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    if col == 'zipcode':
        plt.xticks(rotation=90)
    plt.grid(axis='y')
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Define the list of numerical features to visualize
numerical_features = ['sqft_lot', 'sqft_above', 'sqft_basement', 'yr_built', 'yr_renovated', 'sqft_living15', 'sqft_lot15']

# Iterate through each feature and generate a histogram/KDE plot
for feature in numerical_features:
    plt.figure(figsize=(10, 6))
    sns.histplot(df[feature], kde=True, color='skyblue')
    plt.title(f'Distribution of {feature}')
    plt.xlabel(feature)
    plt.ylabel('Frequency')
    plt.grid(True)
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the correlation matrix for all numerical columns
# Ensure 'price' is included, which is already normalized in 'df'
correlation_matrix = df.select_dtypes(include=['number']).corr()

plt.figure(figsize=(15, 12))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Define lists for key categorical and numerical features
categorical_features_for_plotting = ['bedrooms', 'bathrooms', 'floors', 'waterfront', 'view', 'condition', 'grade']
numerical_features_for_plotting = ['sqft_living', 'sqft_above', 'sqft_living15']

# 2. For each identified categorical feature, create a box plot
for col in categorical_features_for_plotting:
    plt.figure(figsize=(10, 6))
    sns.boxplot(data=df, x=col, y='price', hue=col, palette='viridis', legend=False)
    plt.title(f'Normalized Price Distribution by {col}')
    plt.xlabel(col)
    plt.ylabel('Normalized Price')
    plt.grid(axis='y')
    plt.show()

# 3. For each identified numerical feature, create a scatter plot
for col in numerical_features_for_plotting:
    plt.figure(figsize=(10, 6))
    sns.scatterplot(data=df, x=col, y='price', alpha=0.6)
    plt.title(f'Normalized Price vs. {col}')
    plt.xlabel(col)
    plt.ylabel('Normalized Price')
    plt.grid(True)
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create multi-faceted scatter plots
g = sns.relplot(data=df, x='sqft_living', y='price', col='bedrooms', row='waterfront',
                kind='scatter', height=4, aspect=1.2, facet_kws={'sharex': False, 'sharey': False})
g.fig.suptitle('Normalized Price vs. Square Footage by Bedrooms and Waterfront', y=1.03) # Adjust title position
# plt.tight_layout()
plt.show()
```

## Summary:

### Data Analysis Key Findings

*   **Waterfront Premium**: Properties with a `waterfront` view consistently show a significantly higher average normalized `price` compared to those without, regardless of the number of bedrooms. For instance, a 3-bedroom house without a waterfront has an average normalized price of 0.346186, which increases to 0.598079 for a 3-bedroom house with a waterfront.
*   **Bedroom Impact**: The normalized `price` generally increases with the number of `bedrooms`.
*   **Sqft\_Living and Price Correlation**: Multi-faceted scatter plots show a general positive correlation between `sqft_living` and normalized `price` across different `bedrooms` counts and `waterfront` statuses, indicating that larger living areas are associated with higher prices.

### Insights or Next Steps

*   **Key Price Drivers**: Both `waterfront` status and the number of `bedrooms` are strong determinants of property price, with `waterfront` presence being a particularly significant price multiplier. `sqft_living` also plays a crucial role, with larger homes generally commanding higher prices.
*   **Further Modeling**: It would be beneficial to quantify the interaction effects between `bedrooms`, `waterfront`, and `sqft_living` on `price` using a regression model to better understand their combined influence and predictive power.

```python
price_with_waterfront = average_price_by_waterfront.loc[average_price_by_waterfront['waterfront'] == 1, 'price'].values[0]
price_without_waterfront = average_price_by_waterfront.loc[average_price_by_waterfront['waterfront'] == 0, 'price'].values[0]

price_difference = price_with_waterfront - price_without_waterfront

print(f"Normalized price difference due to waterfront access: {price_difference:.6f}")
```

```python
renovated_houses = df[df['yr_renovated'] > 0]
non_renovated_houses = df[df['yr_renovated'] == 0]

average_price_renovated = renovated_houses['price'].mean()
average_price_non_renovated = non_renovated_houses['price'].mean()

price_difference_renovated = average_price_renovated - average_price_non_renovated

print(f"Average normalized price for renovated houses: {average_price_renovated:.6f}")
print(f"Average normalized price for non-renovated houses: {average_price_non_renovated:.6f}")
print(f"Normalized price difference (Renovated - Non-Renovated): {price_difference_renovated:.6f}")
```
