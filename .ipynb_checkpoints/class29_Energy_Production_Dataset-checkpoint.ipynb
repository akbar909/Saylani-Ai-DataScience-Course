```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Energy Production Dataset.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
sns.set_style("whitegrid")
plt.figure(figsize=(15, 10))
plt.subplot(2, 3, 1)
sns.countplot(x='Source', data=df, hue='Source', palette='plasma', legend=False)
plt.title('Distribution of Source')
plt.xlabel('Source')
plt.ylabel('Count')


plt.subplot(2, 3, 2)
sns.countplot(x='Season', data=df, hue='Season', palette='viridis', legend=False)
plt.title('Distribution of Season')
plt.xlabel('Season')
plt.ylabel('Count')

plt.tight_layout()
plt.show()
```

```python
df['Month'] = df['Month_Name']

plt.figure(figsize=(15, 5))
plt.subplot(1,3,1)
df.groupby('Month')['Production'].sum().plot(kind='bar', color='green')
plt.title('Production by Month')
plt.xlabel('Month')
plt.subplot(1,3,2)
df.groupby('Season')['Production'].sum().plot(kind='bar', color='orange')
plt.title('Production by Season')
plt.xlabel('Season')

plt.subplot(1,3,3)
df.groupby('Source')['Production'].sum().plot(kind='bar', color='blue')
plt.title('Production by Source')
plt.xlabel('Source')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x='Season', y='Production', hue='Source', estimator=sum)
plt.title('Total Energy Production by Source and Season')
plt.ylabel('Cumulative Production')
plt.show()

seasonal_hourly = df.groupby(['Season', 'Start_Hour'])['Production'].mean().reset_index()
winter_summer = seasonal_hourly[seasonal_hourly['Season'].isin(['Winter', 'Summer', 'Spring', 'Fall'])]

plt.figure(figsize=(12, 6))
sns.lineplot(data=winter_summer, x='Start_Hour', y='Production', hue='Season', marker='o')
plt.title('Average Hourly Production Profile: Winter vs Summer vs Sring vs Fall')
plt.xticks(range(0, 24))
plt.grid(True, alpha=0.3)
plt.show()
```
