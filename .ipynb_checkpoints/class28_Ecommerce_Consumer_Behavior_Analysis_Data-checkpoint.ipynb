```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data  = pd.read_csv('Ecommerce_Consumer_Behavior_Analysis_Data.csv')
```

```python
data.head(3)
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
data['Purchase_Amount'] = data['Purchase_Amount'].astype(str).str.replace('$', '', regex=False).str.strip().astype(float)
print(data['Purchase_Amount'].head())
print(data['Purchase_Amount'].dtype)
```

```python
data.describe()
```

```python
x = data['Income_Level']
y= data['Purchase_Amount']
plt.bar(x,y)
plt.xlabel('Income_Level')
plt.ylabel('Purchase_Amount')
plt.title('Purchase_Amount vs Income_Level')
plt.show()
```
