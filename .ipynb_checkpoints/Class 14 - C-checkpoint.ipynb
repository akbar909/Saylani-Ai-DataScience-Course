```python
while True:
    
    minNum = input("Enter Minimum Value: ")
    maxNum = input("Enter Maximum Value: ")

    if not minNum.isdigit() or not maxNum.isdigit():
        print("Enter Valid Input Type, Numbers Only")
        continue
    else:
        minNum = int(minNum)
        maxNum = int(maxNum)
        
    if minNum < maxNum:
        secertNum = random.randrange(minNum, maxNum)
        
        while True:
            userGuess = input("Enter User Guess Number: ")

            if not userGuess.isdigit():
                print("Enter Valid Input, Only Numbers")
                continue
            
            else:
                userGuess = int(userGuess)
            
            if userGuess < secertNum:
                print("You Guess Low Number")
            elif userGuess > secertNum:
                print("You Guess High Number")
            else:
                print("You Guess Correct")
                print("Secert Number is ",secertNum)
                break
    else:
        print("Enter Minimum Value Less Than Maximum Value")
        continue
    break
```

```python
while True:
    chances = 3
    
    minNum = input("Enter Minimum Value: ")
    maxNum = input("Enter Maximum Value: ")

    if not minNum.isdigit() or not maxNum.isdigit():
        print("Enter Valid Input Type, Numbers Only")
        continue
    else:
        minNum = int(minNum)
        maxNum = int(maxNum)
        
    if minNum < maxNum:
        secertNum = random.randrange(minNum, maxNum)
         
        while chances <= 3:
            print(f"You have {chances} Lives to Guess Correct")
            userGuess = input("Enter User Guess Number: ")

            if not userGuess.isdigit():
                print("Enter Valid Input, Only Numbers")
                continue
            
            else:
                userGuess = int(userGuess)
            
            if userGuess < secertNum:
                print("You Guess Low Number")
            elif userGuess > secertNum:
                print("You Guess High Number")
            else:
                print("You Guess Correct")
                print("Secert Number is ",secertNum)
                break
            
            chances -= 1

            if chances == 0:
                print(f"You lost your Lives, Secert Number is {secertNum}")
                break
    else:
        print("Enter Minimum Value Less Than Maximum Value")
        continue
    break
```

```python
print("WELCOME TO MY NUMBER GUESSING GAME")

score = 0 

while True:
    minNum = int(input("Enter a Minimum Value: "))
    maxNum = int(input("Enter a Maximum Value: "))
    
    print(f"You selected Minimum Number {minNum} and Maximum Number {maxNum}")
    
    userLevel = input("Enter Your Difficulty Level:\nEASY\nMEDIUM\nHARD\nIMPOSSIBLE: ").lower()
    
    if userLevel == 'medium':
        chances = 7
        maxNum *= random.randint(2, 10)
    elif userLevel == 'hard':
        chances = 5
        maxNum *= random.randint(12, 45)
    elif userLevel == 'impossible':
        chances = 3
        maxNum *= random.randint(100,1000)
    else:
        chances = 10
        
    print(f"You selected {userLevel} difficulty.\nNow you have {chances} lives to Guess the Correct Number.\nYour Maximum Number now is {maxNum}.")
    
    if minNum < maxNum:
        secretNumber = random.randint(minNum, maxNum)
        
        while chances > 0:
            print(f"You have {chances} lives left to guess correctly.")
            userGuess = input("Enter Your Guess Number: ")

            if not userGuess.isdigit():
                print("Enter Only Numbers, Invalid Input Data Type.")
                continue
            else:
                userGuess = int(userGuess)
                
                if userGuess < secretNumber:
                    print("Your guess is too low.")
                elif userGuess > secretNumber:
                    print("Your guess is too high.")
                else:
                    print("You guessed the correct number!")
                    print(f"Your Tries Count: {score}")
                    break
                    
            chances -= 1
            score += 1
            
            if chances == 0: 
                print(f"You are out of lives. The secret number was {secretNumber}.")
                break

        userChoice = input("Do you want to play again? \nYes\nNo\n").lower()
        if userChoice != 'yes':
            print("You Scared From Me")
            break 
        else:
            continue
    else:
        print("Error: Minimum Value must be less than Maximum Value.")
        continue
    break
```
