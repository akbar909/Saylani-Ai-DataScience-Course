```python
import random
```

```python
word_list = [
    "play", "fun", "game", "win", "lose", "rule", "team", "score", "level", "start",
    "end", "pass", "turn", "card", "dice", "ball", "run", "jump", "talk", "laugh",
    "friend", "name", "person", "man", "boy", "dad", "mom", "brother", "sister", "family",
    "home", "house", "work", "school", "time", "day", "night", "sun", "moon", "star",
    "water", "food", "eat", "drink", "sleep", "sit", "stand", "walk", "see", "look",
    "hear", "sound", "feel", "touch", "think", "know", "learn", "ask", "answer", "help",
    "make", "do", "go", "come", "here", "there", "now", "later", "yes", "no",
    "good", "bad", "happy", "sad", "big", "small", "hot", "cold", "new", "old",
    "fast", "slow", "high", "low", "open", "close", "light", "dark", "full", "empty",
    "ready", "done", "wait", "stop", "move", "find", "lose", "give", "take", "bring",
    "show", "hide", "live", "die", "life", "world", "place", "city", "town", "street",
    "car", "bus", "train", "plane", "boat", "door", "window", "room", "bed", "chair",
    "table", "box", "book", "pen", "paper", "phone", "computer", "money", "love", "like",
    "want", "need", "have", "own", "use", "way", "path", "hope", "wish", "dream",
    "story", "truth", "lie", "kind", "nice", "mean", "busy", "free", "easy", "hard",
    "sure", "maybe", "always", "never", "often", "sometimes", "each", "every", "some", "many"
]
```

```python
print("==== Welcome to word guessing game ====")

while True:
    randomWord = random.choice(word_list).lower()
    print(randomWord)
    
    while True:
        difficultyLevel = input("\nEnter Difficulty (EASY / MEDIUM / HARD) : ").lower()
        
        word = ["-"]*len(randomWord)
        
        if difficultyLevel == "easy":
            chances = len(max(word_list, key=len))+6
            TwoCharToShow = random.sample(range(len(randomWord)),2)
            for i in TwoCharToShow:
                word[i] = randomWord[i]
                
        elif difficultyLevel == "medium":
            chances = len(max(word_list, key=len)) + 3
            characterToShow = random.randint(0, len(randomWord)-1)
            word[characterToShow] = randomWord[characterToShow]

        elif difficultyLevel == "hard":
            chances = len(max(word_list, key=len))
        else:
            print("\nenter valid input (EASY / MEDIUM) / HARD)")
            continue
        break

    print("Word to guess:", "".join(word))
    
    while True:
        hint = input("\nDo you want to hint ? (YES/NO)").lower()
        if hint =='yes':
            revealed_char = random.randint(0, len(randomWord)-1)
            print(f"Hint: {randomWord[revealed_char]} is in word")
            print("".join(word))
            break
            
        elif hint == 'no':
            print("".join(word))
            break
        else:
            print('\nEnter a valid input, (yes/no) only')
        
    
    while chances > 0:
        
        print(f"you have {chances} chances left")
        userInput = input("Enter letter to guess : ").lower()
        
        if len(userInput) > 1 and userInput.isalpha():
            print('Please enter single alphabet character')
            
        if userInput in word and  word.count(userInput) == randomWord.count(userInput):
            print("you already guessed that character")
            print("".join(word))
            continue
            
        elif userInput not in randomWord and len(userInput) < 2:
            print(f"Oops! '{userInput}' character is not in random word")
            chances -= 1 
            
        for i in range(len(randomWord)):
            if randomWord[i] == userInput:
                word[i] = userInput
                chances -= 1
        print("".join(word))
    
        if "-" not in word:
            print(f"Congrulations! You guessed the word : '{randomWord}'")
            break
        if chances == 0:
            print(f"You lost the word was '{randomWord}'")
            
    while True:
        continueGame = input("\nDo you want to play again? (yes/no) ").lower()
        if continueGame == "yes":
            restart = True
            break
        elif continueGame == "no":
            restart = False
            break
        else:
            print('\nPlease enter (yes/no) only')
    if restart:
        continue
    else:
        break
```
