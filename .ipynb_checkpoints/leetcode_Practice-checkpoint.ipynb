### Day 1
#### Two Sum

```python
nums = [2,7,11,15]
target = 9
def twoSum(nums, target):
    
    temp = {}
    for i, values in enumerate(nums):
        diff = target - values
        if diff in temp:
            return [temp[diff], i]
        temp[values] = i
twoSum(nums, target) 
```

### Day 2
#### Roman To integer 

```python
s = "VII"
```

```python
roman_map = {
        'I': 1,'V': 5, 'X': 10, 'L': 50,
        'C': 100, 'D': 500, 'M': 1000
    }
```

```python
n = len(s)
total = 0
```

```python
for i in range(n):
    
    current_value = roman_map[s[i]]
    
    if i + 1 < n and current_value < roman_map[s[i+1]]:
        total -= current_value
    else:
        total += current_value
                
print(total)
```

### Day 4
#### Valid Parentheses

```python
s = "(){}[]"
stack = []
mapping = {
')' : '(',
']' : '[',
'}' : '{'
}
```

```python
def isValid(s):
    for char in s:
        if char in mapping.values():
            stack.append(char)
        elif char in mapping:
            if not stack or stack[-1] != mapping[char]:
                return False
            stack.pop()
    return not stack
```

```python
isValid(s)
```

### Day 5
#### 1. Remove Duplicates from Sorted Array

```python
# 1. Remove Duplicates from Sorted Array
nums = [1,1,2]
```

```python
def removeDuplicates(nums):
    if not nums:
        return 0
        
    k = 1  
    for i in range(1, len(nums)):
        if nums[i] != nums[k - 1]:
            nums[k] = nums[i]
            k += 1

    return k
```

```python
removeDuplicates(nums)
```

### Day 7 
#### Find the Index of the First Occurrence in a String

```python
haystack = "leetcode"
needle = "code"
```

```python
def strStr(haystack, needle):
    if needle == "":
        return 0
    
    n, m = len(haystack), len(needle)
    
    for i in range(n - m + 1):
        if haystack[i:i+m] == needle:
            return i
    return -1
```

```python
strStr(haystack, needle)
```

### Day 8
#### Search Insert Position

```python
nums = [1,3,5,6]
target = 5
```

```python
def searchInsert(nums,target):
    for i in range(len(nums)):
        if target == nums[i]:
            return i
        elif target < nums[i]:
            nums[i] = target
            return i
    return i+1
```

```python
searchInsert(nums, target)
```

#### 2. Plus One

```python
digits = [9]
```

```python
def plusOne(digits):
    for i in range(len(digits) -1, -1, -1):
        if digits[i] < 9:
            digits[i] += 1
            return digits
        else:
            digits[i] = 0
    return [1]+ digits
plusOne(digits)
```

### Day 10
#### Add Binary

```python
a = "1010"
b = "1011"
```

```python
result = []
i, j = len(a) - 1, len(b) - 1
carry = 0
```

```python
def addBinary(a, b):
        result = []
        i, j = len(a) - 1, len(b) - 1
        carry = 0

        while i >= 0 or j >= 0 or carry:
            current_sum = carry

            if i >= 0:
                current_sum += int(a[i])
                i -= 1
            if j >= 0:
                current_sum += int(b[j])
                j -= 1

            result.append(str(current_sum % 2))
            carry = current_sum // 2

        return "".join(reversed(result))
```

```python
addBinary(a,b)
```

### Day 11
#### Sqrt(x) :  return the square root of x rounded down to the nearest integer. The returned integer should be non-negative as well.

```python
x = 8
```

```python
def mySqrt(x):
    if x < 2:
        return x
    left , right = 1, x//2

    while left <= right:
        mid = (left + right)//2
        if mid * mid == x:
            return mid
        elif mid * mid < x:
            left = mid +1
        else:
            right = mid -1
    return right
```

```python
mySqrt(x)
```

### Day 12
#### Climbing Stairs

```python
n = 5
```

```python
def climbStairs(n):
        if n <= 2:
            return n

        first, second = 1, 2
        for i in range(3, n + 1):
            third = first + second
            first = second
            second = third
        return second
```

```python
climbStairs(n)
```

### Day 13 : Merge Sorted Array

```python
nums1 = [1,2,3,0,0,0] 
m = 3
nums2 = [2,5,6]
n = 3
```

```python
def merge(nums1, m, nums2, n):
    i = m - 1 
    j = n - 1  
    k = m + n - 1 

    while i >= 0 and j >= 0:
        if nums1[i] > nums2[j]:
            nums1[k] = nums1[i]
            i -= 1
        else:
            nums1[k] = nums2[j]
            j -= 1
        k -= 1

    while j >= 0:
        nums1[k] = nums2[j]
        j -= 1
        k -= 1

    return nums1
```

```python
merge(nums1,m, nums2,n)
```

### Day 14 : Longest Substring Without Repeating Characters

```python
s = "abcabcbb"
```

```python
def lengthOfLongestSubstring(s):
    charSet = set()
    left = 0
    result = 0
    for right in range(len(s)):
        while s[right] in charSet:
            charSet.remove(s[left])
            left += 1
        charSet.add(s[right])
        result = max(result, len(charSet))
    
    return result   
```

```python
lengthOfLongestSubstring(s)
```
