# TUPLES : ordered, unchangeable, indexed and allow duplicate values.

```python
myTuple = (1,2,3,4,5,6,7,8,9,0)
```

```python
type(myTuple)
```

```python
len(myTuple)
```

```python
myTuple[1]
```

```python
myTuple[::-1]
```

```python
myTuple.count(9)
```

```python
myTuple.index(0)
```

# SETS : unordered, unchangeable*, and unindexed.. Duplicates Not Allowed
# * Note: Set items are unchangeable, but you can remove items and add new items.

```python
myset = {"A", "B",1, True}
type(myset)
```

```python
myset
```

```python
myset.add("add item")
myset
```

```python
newItems = {"new 1", "new2 "}
myset.update(newItems)
myset
```
