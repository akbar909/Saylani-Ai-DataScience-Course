```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('zomato.csv', encoding='latin-1')
df.head(3)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.dropna(subset=['Cuisines'], inplace=True)
```

```python
df.duplicated().sum()
```

### CATEGORICAL COLUMNS DISTRIBUTION 

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Distribution of Categorical Columns', fontsize=16, fontweight='bold')

# 1. Rating Text Distribution
rating_counts = df['Rating text'].value_counts()
colors = ['#1abc9c', '#27ae60', '#f39c12', '#e74c3c', '#8e44ad', '#2c3e50']
axes[0, 0].bar(rating_counts.index, rating_counts.values, color=colors[:len(rating_counts)])
axes[0, 0].set_title('Rating Text Distribution')
axes[0, 0].set_xlabel('Rating Category')
axes[0, 0].set_ylabel('Frequency')
axes[0, 0].tick_params(axis='x', rotation=45)

# 2. Has Online Delivery (Pie Chart)
delivery_counts = df['Has Online delivery'].value_counts()
axes[0, 1].pie(delivery_counts, labels=delivery_counts.index, autopct='%1.1f%%', 
               colors=['#e74c3c', '#2ecc71'], explode=[0.05, 0])
axes[0, 1].set_title('Has Online Delivery')

# 3. Has Table Booking (Pie Chart)
booking_counts = df['Has Table booking'].value_counts()
axes[1, 0].pie(booking_counts, labels=booking_counts.index, autopct='%1.1f%%', 
               colors=['#3498db', '#f1c40f'], explode=[0.05, 0])
axes[1, 0].set_title('Has Table Booking')

# 4. Top 10 Cities
city_counts = df['City'].value_counts().head(10)
axes[1, 1].barh(city_counts.index, city_counts.values, color=plt.cm.viridis(range(0, 256, 25)))
axes[1, 1].set_title('Top 10 Cities by Restaurant Count')
axes[1, 1].set_xlabel('Number of Restaurants')
axes[1, 1].invert_yaxis()

plt.tight_layout()
plt.show()

# =====================================================
# FREQUENCY TABLES FOR CATEGORICAL COLUMNS
# =====================================================
print('=' * 60)
print('RATING TEXT FREQUENCY')
print('=' * 60)
print(df['Rating text'].value_counts())

print('\n' + '=' * 60)
print('ONLINE DELIVERY FREQUENCY')
print('=' * 60)
print(df['Has Online delivery'].value_counts())

print('\n' + '=' * 60)
print('TABLE BOOKING FREQUENCY')
print('=' * 60)
print(df['Has Table booking'].value_counts())

print('\n' + '=' * 60)
print('TOP 15 CITIES BY RESTAURANT COUNT')
print('=' * 60)
print(df['City'].value_counts().head(15))

# %%
# =====================================
# RATING DISTRIBUTION BY PRICE RANGE
# =====================================
fig2, ax2 = plt.subplots(figsize=(10, 6))

fig2.suptitle('Rating Analysis by Price Range', fontsize=14, fontweight='bold')

# Box Plot
# Note: We pass ax=ax2 (the single axis we just created)
sns.boxplot(x='Price range', y='Aggregate rating', data=df, ax=ax2, palette='viridis')

ax2.set_title('Rating by Price Range (Box Plot)')
ax2.set_xlabel('Price Range')
ax2.set_ylabel('Aggregate Rating')

plt.tight_layout()
plt.show()
```

### GEOGRAPHIC / CITY-WISE ANALYSIS

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Geographic Analysis', fontsize=16, fontweight='bold')

# 1. Top 15 Cities by Average Rating
city_rating = df.groupby('City')['Aggregate rating'].mean().sort_values(ascending=False).head(15)
axes[0, 0].barh(city_rating.index, city_rating.values, color=plt.cm.RdYlGn(range(50, 256, 13)))
axes[0, 0].set_title('Top 15 Cities by Average Rating')
axes[0, 0].set_xlabel('Average Rating')
axes[0, 0].invert_yaxis()

# 2. Top 15 Cities by Total Votes
city_votes = df.groupby('City')['Votes'].sum().sort_values(ascending=False).head(15)
axes[0, 1].barh(city_votes.index, city_votes.values, color=plt.cm.Blues(range(100, 256, 10)))
axes[0, 1].set_title('Top 15 Cities by Total Votes')
axes[0, 1].set_xlabel('Total Votes')
axes[0, 1].invert_yaxis()

# 3. Average Cost by City (Top 15 by restaurant count)
top_cities = df['City'].value_counts().head(15).index
city_cost = df[df['City'].isin(top_cities)].groupby('City')['Average Cost for two'].mean().sort_values(ascending=False)
axes[1, 0].barh(city_cost.index, city_cost.values, color=plt.cm.Oranges(range(100, 256, 10)))
axes[1, 0].set_title('Average Cost for Two by City (Top 15 Cities)')
axes[1, 0].set_xlabel('Average Cost')
axes[1, 0].invert_yaxis()

fig.delaxes(axes[1, 1])

plt.tight_layout()
plt.show()

# City-wise Summary Statistics
print('=' * 60)
print('CITY-WISE SUMMARY (Top 10 Cities by Restaurant Count)')
print('=' * 60)
city_summary = df.groupby('City').agg({
    'Restaurant ID': 'count',
    'Aggregate rating': 'mean',
    'Votes': 'sum',
    'Average Cost for two': 'mean'
}).rename(columns={'Restaurant ID': 'Count'}).sort_values('Count', ascending=False).head(10)
print(city_summary.round(2))
```

### BUSINESS INSIGHTS (Delivery & Booking Impact)

```python
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle('Business Insights Analysis', fontsize=16, fontweight='bold')

# 1. Online Delivery vs Average Rating
delivery_rating = df.groupby('Has Online delivery')['Aggregate rating'].mean()
axes[0, 0].bar(delivery_rating.index, delivery_rating.values, color=['#e74c3c', '#2ecc71'])
axes[0, 0].set_title('Average Rating by Online Delivery')
axes[0, 0].set_xlabel('Has Online Delivery')
axes[0, 0].set_ylabel('Average Rating')
axes[0, 0].set_ylim(0, 5)

# 2. Table Booking vs Average Rating
booking_rating = df.groupby('Has Table booking')['Aggregate rating'].mean()
axes[0, 1].bar(booking_rating.index, booking_rating.values, color=['#3498db', '#f1c40f'])
axes[0, 1].set_title('Average Rating by Table Booking')
axes[0, 1].set_xlabel('Has Table Booking')
axes[0, 1].set_ylabel('Average Rating')
axes[0, 1].set_ylim(0, 5)

# 3. Online Delivery vs Average Votes
delivery_votes = df.groupby('Has Online delivery')['Votes'].mean()
axes[0, 2].bar(delivery_votes.index, delivery_votes.values, color=['#e74c3c', '#2ecc71'])
axes[0, 2].set_title('Average Votes by Online Delivery')
axes[0, 2].set_xlabel('Has Online Delivery')
axes[0, 2].set_ylabel('Average Votes')

# 4. Price Range vs Online Delivery (Stacked Bar)
price_delivery = pd.crosstab(df['Price range'], df['Has Online delivery'], normalize='index') * 100
price_delivery.plot(kind='bar', stacked=True, ax=axes[1, 0], color=['#e74c3c', '#2ecc71'])
axes[1, 0].set_title('Online Delivery % by Price Range')
axes[1, 0].set_xlabel('Price Range')
axes[1, 0].set_ylabel('Percentage')
axes[1, 0].legend(title='Online Delivery')
axes[1, 0].tick_params(axis='x', rotation=0)

# 5. Price Range vs Table Booking (Stacked Bar)
price_booking = pd.crosstab(df['Price range'], df['Has Table booking'], normalize='index') * 100
price_booking.plot(kind='bar', stacked=True, ax=axes[1, 1], color=['#3498db', '#f1c40f'])
axes[1, 1].set_title('Table Booking % by Price Range')
axes[1, 1].set_xlabel('Price Range')
axes[1, 1].set_ylabel('Percentage')
axes[1, 1].legend(title='Table Booking')
axes[1, 1].tick_params(axis='x', rotation=0)

# 6. Rating Distribution: With vs Without Online Delivery
sns.boxplot(x='Has Online delivery', y='Aggregate rating', data=df, ax=axes[1, 2], palette=['#e74c3c', '#2ecc71'])
axes[1, 2].set_title('Rating Distribution by Online Delivery')
axes[1, 2].set_xlabel('Has Online Delivery')
axes[1, 2].set_ylabel('Aggregate Rating')

plt.tight_layout()
plt.show()

# Business Insights Summary
print('=' * 60)
print('BUSINESS INSIGHTS SUMMARY')
print('=' * 60)

print('\n ONLINE DELIVERY IMPACT:')
print(df.groupby('Has Online delivery').agg({
    'Aggregate rating': 'mean',
    'Votes': 'mean',
    'Average Cost for two': 'mean'
}).round(2))

print('\n TABLE BOOKING IMPACT:')
print(df.groupby('Has Table booking').agg({
    'Aggregate rating': 'mean',
    'Votes': 'mean',
    'Average Cost for two': 'mean'
}).round(2))

print('\n PRICE RANGE vs SERVICES:')
print(pd.crosstab(df['Price range'], df['Has Online delivery'], margins=True))
```
