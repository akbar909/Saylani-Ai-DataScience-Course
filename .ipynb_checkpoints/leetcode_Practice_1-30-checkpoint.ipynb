### Day 1
#### Two Sum

```python
nums = [2,7,11,15]
target = 13
def twoSum(nums, target):
    
    temp = {}
    for i, values in enumerate(nums):
        diff = target - values
        if diff in temp:
            return [temp[diff], i]
        temp[values] = i
twoSum(nums, target) 
```

### Day 4
#### Valid Parentheses

```python
s = "(){}[]"
stack = []
mapping = {
')' : '(',
']' : '[',
'}' : '{'
}
```

```python
def isValid(s):
    for char in s:
        if char in mapping.values():
            stack.append(char)
        elif char in mapping:
            if not stack or stack[-1] != mapping[char]:
                return False
            stack.pop()
    return not stack
```

```python
isValid(s)
```

### Day 5
#### 1. Remove Duplicates from Sorted Array

```python
# 1. Remove Duplicates from Sorted Array
nums = [1,1,2]
```

```python
def removeDuplicates(nums):
    if not nums:
        return 0
        
    k = 1  
    for i in range(1, len(nums)):
        if nums[i] != nums[k - 1]:
            nums[k] = nums[i]
            k += 1

    return k
```

```python
removeDuplicates(nums)
```

### Day 7 
#### Find the Index of the First Occurrence in a String

```python
haystack = "leetcode"
needle = "code"
```

```python
def strStr(haystack, needle):
    if needle == "":
        return 0
    
    n, m = len(haystack), len(needle)
    
    for i in range(n - m + 1):
        if haystack[i:i+m] == needle:
            return i
    return -1
```

```python
nums = [1,3,5,6]
target = 5
```

```python
def searchInsert(nums,target):
    for i in range(len(nums)):
        if target == nums[i]:
            return i
        elif target < nums[i]:
            nums[i] = target
            return i
    return i+1
```

#### 2. Plus One

```python
digits = [9]
```

```python
def plusOne(digits):
    for i in range(len(digits) -1, -1, -1):
        if digits[i] < 9:
            digits[i] += 1
            return digits
        else:
            digits[i] = 0
    return [1]+ digits
plusOne(digits)
```

### Day 10
#### Add Binary

```python
a = "1010"
b = "1011"
```

```python
result = []
i, j = len(a) - 1, len(b) - 1
carry = 0
```

```python
def addBinary(a, b):
        result = []
        i, j = len(a) - 1, len(b) - 1
        carry = 0

        while i >= 0 or j >= 0 or carry:
            current_sum = carry

            if i >= 0:
                current_sum += int(a[i])
                i -= 1
            if j >= 0:
                current_sum += int(b[j])
                j -= 1

            result.append(str(current_sum % 2))
            carry = current_sum // 2

        return "".join(reversed(result))
```

```python
addBinary(a,b)
```

### Day 11
#### Sqrt(x) :  return the square root of x rounded down to the nearest integer. The returned integer should be non-negative as well.

```python
x = 8
```

```python
def mySqrt(x):
    if x < 2:
        return x
    left , right = 1, x//2

    while left <= right:
        mid = (left + right)//2
        if mid * mid == x:
            return mid
        elif mid * mid < x:
            left = mid +1
        else:
            right = mid -1
    return right
```

```python
mySqrt(x)
```

### Day 12
#### Climbing Stairs

```python
n = 5
```

```python
def climbStairs(n):
        if n <= 2:
            return n

        first, second = 1, 2
        for i in range(3, n + 1):
            third = first + second
            first = second
            second = third
        return second
```

```python
climbStairs(n)
```

### Day 13 : Merge Sorted Array

```python
nums1 = [1,2,3,0,0,0] 
m = 3
nums2 = [2,5,6]
n = 3
```

```python
def merge(nums1, m, nums2, n):
    i = m - 1 
    j = n - 1  
    k = m + n - 1 

    while i >= 0 and j >= 0:
        if nums1[i] > nums2[j]:
            nums1[k] = nums1[i]
            i -= 1
        else:
            nums1[k] = nums2[j]
            j -= 1
        k -= 1

    while j >= 0:
        nums1[k] = nums2[j]
        j -= 1
        k -= 1

    return nums1
```

```python
merge(nums1,m, nums2,n)
```

### Day 14 : Longest Substring Without Repeating Characters

```python
s = "abcabcbb"
```

```python
def lengthOfLongestSubstring(s):
    charSet = set()
    left = 0
    result = 0
    for right in range(len(s)):
        while s[right] in charSet:
            charSet.remove(s[left])
            left += 1
        charSet.add(s[right])
        result = max(result, len(charSet))
    
    return result   
```

```python
lengthOfLongestSubstring(s)
```

#### Day 16: Integer to Roman

```python
num = 3749
```

```python
def intToRoman(num):
    data = [
            ["I", 1], ["IV", 4], ["V", 5], ["IX", 9],
            ["X", 10], ["XL", 40], ["L", 50], ["XC", 90],
            ["C", 100], ["CD", 400], ["D", 500], ["CM", 900],
            ["M", 1000]
        ]
    result = ""
    for data, val in reversed(data):
        if num // val:
            count = num // val
            result += (data * count)
            num = num % val
    return result
```

```python
intToRoman(num)
```

#### Day 17 : Rotate Image

```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]
```

```python
def rotate(matrix) -> None:
    l, r = 0, len(matrix) - 1
    while l < r:
        for i in range(r - l):
            top, bottom = l, r
            topLeft = matrix[top][l + i]
            
            matrix[top][l + i] = matrix[bottom - i][l]
            matrix[bottom - i][l] = matrix[bottom][r - i]
            matrix[bottom][r - i] = matrix[top + i][r]
            matrix[top + i][r] = topLeft

        r -= 1
        l += 1
```

```python
rotate(matrix)
```

#### Day 18 : Reverse Integer

```python
x = -123
```

```python
def reverse(x):
    
    x = str(x)

    if x[0] == "-":
        reversedNum = "-"+ x[:0:-1]
    else:
        reversedNum = x[::-1]

    reversedNum = int(reversedNum)
    if reversedNum < -2**31 or reversedNum > 2**31-1:
        return 0
    return reversedNum
    
reverse(x)
```

#### Day 19 : Letter Combinations of a Phone Number

```python
digits = "23"
```

```python
def letterCombinations(digits):
        res = []
        digitToChar = { 
            "2": "abc",
            "3": "def",
            "4": "ghi",
            "5": "jkl",
            "6": "mno",
            "7": "pqrs",
            "8": "tuv",
            "9": "wxyz" 
        }

        def backtrack(i, curStr):
            if len(curStr) == len(digits):
                res.append(curStr)
                return
            for c in digitToChar[digits[i]]:
                backtrack(i + 1, curStr + c)

        if digits:
            backtrack(0, "")

        return res
```

```python
letterCombinations(digits)
```

### Day 20 : Generate Parentheses

```python
n = 3
```

```python
def generateParenthesis(n):
    stack = []
    res = []
    
    def backtrack(openN, closedN):
        if openN == closedN == n:
            res.append("".join(stack))
            return
    
        if openN < n:
            stack.append("(")
            backtrack(openN + 1, closedN)
            stack.pop()
    
        if closedN < openN:
            stack.append(")")
            backtrack(openN, closedN + 1)
            stack.pop()
    
    backtrack(0, 0)
    return res
```

```python
generateParenthesis(n)
```

#### Day 21 : Group Anagrams

```python
strs = ["eat","tea","tan","ate","nat","bat"]
```

```python
def groupAnagrams(strs):
    res = {}
    for s in strs:
        key = ''.join(sorted(s)) 
        if key not in res:
            res[key] = []
        res[key].append(s)
        
    return list(res.values())
```

```python
groupAnagrams(strs)
```

### Day 22 : String to Integer (atoi)

```python
s = "1337c0d3"
```

```python
def myAtoi(s):
    s = s.strip()
    res = ""
    for i, ch in enumerate(s):
        if ch in "+-":
            if i == 0:      
                res += ch
            else:
                break
        elif ch.isdigit():
            res += ch
        else:
            break
    if res == "" or res == "+" or res == "-":
            return 0
    num = int(res)
    if num < -2**31:  
        return -2**31
    elif num > 2**31 - 1:
        return 2**31 - 1

    return num
```

```python
myAtoi(s)
```

#### Day 23 : Find First and Last Position of Element in Sorted Array

```python
nums = [5,7,7,8,8,1]
target = 8
```

```python
def searchRange(nums , target):
    result = []
    if not nums:
        return [-1, -1]

    for i in range(len(nums)):
        if nums[i] == target:
            result.append(i)

    if not result:
        return [-1, -1]

    if len(result) == 1:
        return [result[0], result[0]]

    return [result[0], result[-1]]
```

```python
searchRange(nums, target)
```

### Day 24 : Count and Say

```python
n = 4
```

```python
def countAndSay(n):
    if n == 1:
        return "1"
        
    prev=countAndSay(n-1)
    result = ""
    count = 1
    
    for i in range(len(prev)):
        if i + 1 < len(prev) and prev[i] == prev[i + 1]:
            count += 1
        else:
            result += str(count) + prev[i]
            count = 1
    return result
```

```python
countAndSay(n)
```

### Day 25 : Search in Rotated Sorted Array

```python
nums = [4,5,6,7,0,1,2]
target = 0
```

```python
def search(nums, target):
    l, r = 0, len(nums) - 1

    while l <= r:
        mid = (l + r) // 2

        # Check if target found
        if target == nums[mid]:
            return mid

        # Left half is sorted
        if nums[l] <= nums[mid]:
            if nums[l] <= target < nums[mid]:
                r = mid - 1
            else:
                l = mid + 1
        # Right half is sorted
        else:
            if nums[mid] < target <= nums[r]:
                l = mid + 1
            else:
                r = mid - 1

    return -1
```

```python
search(nums,target)
```

# Day 26 : Pow(x, n)

```python
x = 2.00000
n = 10
```

```python
def myPow(x, n):
    def helper(x, n):
        if x == 0:
            return 0
        if n == 0:
            return 1
        
        res = helper(x * x, n // 2)
        return x * res if n % 2 else res
    
    res = helper(x, abs(n))
    return res if n >= 0 else 1 / res
```

```python
myPow(x,n)
```

### Day27: Longest Palindromic Substring

```python
 s = "babad"
```

```python
def longestPalindrome(s):
    res = ""
    resLen = 0

    for i in range(len(s)):
        # odd length
        l, r = i, i
        while l >= 0 and r < len(s) and s[l] == s[r]:
            if (r - l + 1) > resLen:
                res = s[l:r+1]
                resLen = r - l + 1
            l -= 1
            r += 1
        # even length
        l, r = i, i + 1
        while l >= 0 and r < len(s) and s[l] == s[r]:
            if (r - l + 1) > resLen:
                res = s[l:r+1]
                resLen = r - l + 1
            l -= 1
            r += 1

    return res
```

```python
longestPalindrome(s)
```

### Day 28 : Multiply Strings

```python
num1 = "2"
num2 = "3"
```

```python
def multiply(num1, num2):
    if "0" in [num1, num2]:
        return "0"
    
    res = [0] * (len(num1) + len(num2))
    num1, num2 = num1[::-1], num2[::-1]
    
    for i1 in range(len(num1)):
        for i2 in range(len(num2)):
            digit = int(num1[i1]) * int(num2[i2])
            res[i1 + i2] += digit
            res[i1 + i2 + 1] += (res[i1 + i2] // 10)
            res[i1 + i2] = res[i1 + i2] % 10
            
    res, beg = res[::-1], 0
    while beg < len(res) and res[beg] == 0:
        beg += 1
        
    res = map(str, res[beg:])
    return "".join(res)
```

```python
multiply(num1,num2)
```
