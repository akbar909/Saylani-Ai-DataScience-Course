```python
Fruits = ["apple", "banana", "cherry"]
i = 0
while i < len(Fruits):
    print(Fruits[i])
    i = i+1
```

```python
Fruits = ["apple", "banana", "cherry", "peach", "cherry","pomegranate","lemon"]
i = 0
while i < len(Fruits):
    print(Fruits[i])
    i = i+1
```
