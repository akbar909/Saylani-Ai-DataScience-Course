```python
import matplotlib.pyplot as plt
```

```python
age_gender_counts = df.groupby(['driver_age', 'driver_gender']).size().unstack()

age_gender_counts.plot(kind='bar')
plt.xlabel('Driver Age')
plt.ylabel('Count')
plt.title('Driver Counts by Age and Gender')
plt.show()
```

```python
gender_numeric = df['driver_gender'].map({'M': 0, 'F': 1})

plt.scatter(df['driver_age'], gender_numeric)
plt.yticks([0,1], ['M','F'])
plt.show()
```

```python
df['driver_gender'].value_counts().plot(kind='bar')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```
