```python
import random
```

```python
def toss_coin():
    return random.choice(["Heads", "Tails"])

results = [toss_coin() for i in range(1000)]

print("P(Heads): ", results.count("Heads") / len(results))
print("P(Tails): ", results.count("Tails") / len(results))
```

```python
def playing_cards():
    suits = ['Hearts', 'Diamonds', 'Clubs', 'Spades']
    ranks = ['2', '3', '4', '5', '6', '7', '8', '9', '10', 'Jack', 'Queen', 'King', 'Ace']
    random_suit = random.choice(suits)
    random_rank = random.choice(ranks)
    return f"{random_rank} of {random_suit}"
results = [playing_cards() for _ in range(52)]
ace_count = sum(1 for card in results if 'Ace' in card)
# print(results)
print('P(ACE) : ', results.count('Ace of Spades')/ len(results))
```

```python
def roll_dice():
    return random.randint(1,6)

results = [roll_dice() for i in range(1)]
print('P(6)', results.count(6)/len(results))
```

```python
def toss_coin():
    coin1 = random.choice(["H", "T"])
    coin2 = random.choice(["H", "T"])
    coin3 = random.choice(["H", "T"])
    return coin1, coin2, coin3

results = [toss_coin() for i in range(8)]
print(results)
print('P(TTT): ', results.count(('H', 'T', 'T'))/ len(results))
```
