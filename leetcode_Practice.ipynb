### Day 1
#### Two Sum

```python
nums = [2,7,11,15],
target = 9
temp = {}
for i, values in enumerate(nums):
    diff = target - values
    if diff in temp:
        print( [temp[diff], i] )
    temp[values] = i
```

### Day 2
#### Roman To integer 

```python
s = "VII"
```

```python
 roman_map = {
            'I': 1,'V': 5, 'X': 10, 'L': 50,
            'C': 100, 'D': 500, 'M': 1000
        }
```

```python
n = len(s)
total = 0
```

```python
for i in range(n):
    
    current_value = roman_map[s[i]]
    
    if i + 1 < n and current_value < roman_map[s[i+1]]:
        total -= current_value
    else:
        total += current_value
                
print(total)
```

### Day 3
#### Longest Common Prefix

```python
strs = ["flower","flow","flight"]
```

```python
def longestCommonPrefix(strs):
    for i in range(len(strs[0])):
        
        for string in strs[1:]:
            
            if len(string) <= i or string[i] != strs[0][i]:
                
                return strs[0][:i]
    
    return strs[0]
```

```python
longestCommonPrefix(strs)
```

### Day 4
#### Valid Parentheses

```python
 s = "(){}[]"
stack = []
mapping = {
    ')' : '(',
    ']' : '[',
    '}' : '{'
}
```

```python
def isValid(s):
    for char in s:
        if char in mapping.values():
            stack.append(char)
        elif char in mapping:
            if not stack or stack[-1] != mapping[char]:
                return False
            stack.pop()
    return not stack
```

```python
isValid(s)
```
