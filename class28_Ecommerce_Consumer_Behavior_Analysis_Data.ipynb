## Distributions

```python
# Set aesthetic style for the plots
sns.set_style("whitegrid")

# Set up the figure size for better readability
plt.figure(figsize=(15, 10))

# 1. Distribution of 'Age'
plt.subplot(2, 3, 1)
sns.histplot(df['Age'], bins=10, kde=True)
plt.title('Distribution of Customer Age')
plt.xlabel('Age')
plt.ylabel('Count')

# 2. Distribution of 'Gender'
plt.subplot(2, 3, 2)
sns.countplot(x='Gender', data=df, hue='Gender', palette='viridis', legend=False)
plt.title('Distribution of Customer Gender')
plt.xlabel('Gender')
plt.ylabel('Count')

# 3. Distribution of 'Income_Level'
plt.subplot(2, 3, 3)
sns.countplot(x='Income_Level', data=df, hue='Income_Level', palette='magma', legend=False)
plt.title('Distribution of Customer Income Level')
plt.xlabel('Income Level')
plt.ylabel('Count')

# 4. Distribution of 'Education_Level'
plt.subplot(2, 3, 4)
sns.countplot(x='Education_Level', data=df, hue='Education_Level', palette='plasma', legend=False)
plt.title('Distribution of Customer Education Level')
plt.xlabel('Education Level')
plt.ylabel('Count')

# 5. Distribution of 'Frequency_of_Purchase'
plt.subplot(2, 3, 5)
sns.histplot(df['Frequency_of_Purchase'], bins=10, kde=True)
plt.title('Distribution of Purchase Frequency')
plt.xlabel('Frequency of Purchase')
plt.ylabel('Count')

# 6. Distribution of 'Purchase_Amount'
plt.subplot(2, 3, 6)
sns.histplot(df['Purchase_Amount'], bins=10, kde=True)
plt.title('Distribution of Purchase Amount')
plt.xlabel('Purchase Amount')
plt.ylabel('Count')

plt.tight_layout()
plt.show()
```

## 1. Time-Based Analysis
Analyzes when customers make purchases - by day of week and month. Helps identify peak shopping days and seasonal trends to schedule promotions, staff resources, and manage inventory effectively.

```python
# Extract time features
df['Day'] = df['Time_of_Purchase'].dt.day_name()
df['Month'] = df['Time_of_Purchase'].dt.month_name()

# Visualize patterns
plt.figure(figsize=(15, 5))

plt.subplot(1, 2, 1)
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
df['Day'] = pd.Categorical(df['Day'], categories=day_order, ordered=True)
df['Day'].value_counts().sort_index().plot(kind='bar', color='coral')
plt.title('Number of Purchases by Day')
plt.xlabel('Day of Week')

plt.subplot(1, 2, 2)
df.groupby('Month')['Purchase_Amount'].sum().plot(kind='bar', color='green')
plt.title('Total Revenue by Month')
plt.xlabel('Month')

plt.tight_layout()
plt.show()
```

## 2. Discount Impact Analysis
Compares purchasing behavior with and without discounts. Shows if discounts actually increase sales or just reduce profit margins, helping optimize pricing strategies.

```python
# Compare discount vs non-discount purchases
discount_comparison = df.groupby('Discount_Used').agg({
    'Purchase_Amount': ['mean', 'sum', 'count'],
    'Frequency_of_Purchase': 'mean'
}).round(2)
print("Discount Impact:")
print(discount_comparison)

# Visualize
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.boxplot(x='Discount_Used', y='Purchase_Amount', data=df, hue='Discount_Used', palette='Set2', legend=False)
plt.title('Purchase Distribution: Discount vs No Discount')

plt.subplot(1, 2, 2)
df.groupby('Discount_Used')['Purchase_Amount'].mean().plot(kind='bar', color=['red', 'green'])
plt.title('Average Purchase Amount')
plt.ylabel('Amount ($)')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

## 3. Social Media & Ad Engagement Impact
Measures how social media presence and advertisements influence customer spending. Determines which marketing channels deliver the best return on investment.

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
sns.boxplot(x='Social_Media_Influence', y='Purchase_Amount', data=df, hue='Social_Media_Influence', palette='Blues', legend=False)
plt.title('Social Media Influence on Spending')

plt.subplot(1, 3, 2)
sns.boxplot(x='Engagement_with_Ads', y='Purchase_Amount', data=df, hue='Engagement_with_Ads', palette='Greens', legend=False)
plt.title('Ad Engagement Impact')

plt.subplot(1, 3, 3)
# Combined effect
combined = df.groupby(['Social_Media_Influence', 'Engagement_with_Ads'])['Purchase_Amount'].mean().unstack()
sns.heatmap(combined, annot=True, fmt='.0f', cmap='YlOrRd')
plt.title('Combined Marketing Effect')

plt.tight_layout()
plt.show()
```

## 4. Customer Segmentation
Groups customers by income and education level to identify which segments spend the most. Enables targeted marketing campaigns for different customer types.

```python
# Create customer segments
segment_analysis = df.groupby(['Income_Level', 'Education_Level']).agg({
    'Purchase_Amount': 'mean',
    'Frequency_of_Purchase': 'mean',
    'Customer_ID': 'count'
}).round(2)
segment_analysis.columns = ['Avg_Purchase', 'Avg_Frequency', 'Customer_Count']
print("\nCustomer Segments:")
print(segment_analysis)

# Heatmap visualization
plt.figure(figsize=(10, 6))
pivot = df.pivot_table(values='Purchase_Amount', 
                        index='Income_Level', 
                        columns='Education_Level', 
                        aggfunc='mean')
sns.heatmap(pivot, annot=True, fmt='.0f', cmap='RdYlGn', linewidths=0.5)
plt.title('Average Purchase Amount by Income & Education')
plt.show()
```

## 5. Correlation Analysis
Identifies relationships between variables like age, purchase amount, and frequency. Helps predict customer behavior and understand what drives purchases.

```python
# Select numeric columns
numeric_cols = ['Age', 'Frequency_of_Purchase', 'Purchase_Amount']
correlation_matrix = df[numeric_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, 
            square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title('Correlation Between Numeric Variables')
plt.show()

# Print insights
print("\nCorrelation Insights:")
print(correlation_matrix)
```

## 6. High-Value Customer Identification
Finds the top 10% of customers who generate the most revenue. Allows focusing retention efforts and VIP programs on the most profitable customers.

```python
# Calculate customer lifetime value
df['Customer_Value'] = df['Purchase_Amount'] * df['Frequency_of_Purchase']

# Identify top 10% customers
top_percentile = df['Customer_Value'].quantile(0.90)
high_value_customers = df[df['Customer_Value'] >= top_percentile]

print(f"\nHigh-Value Customers (Top 10%): {len(high_value_customers)}")
print(f"They contribute: ${high_value_customers['Purchase_Amount'].sum():,.2f}")

# Profile of high-value customers
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
high_value_customers['Gender'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('High-Value Customers by Gender')

plt.subplot(1, 3, 2)
high_value_customers['Income_Level'].value_counts().plot(kind='bar', color='gold')
plt.title('High-Value Customers by Income')
plt.xticks(rotation=45)

plt.subplot(1, 3, 3)
high_value_customers['Education_Level'].value_counts().plot(kind='bar', color='purple')
plt.title('High-Value Customers by Education')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

## 7. Age Group Analysis
Examines purchasing patterns across different age ranges (18-25, 26-35, 36-45, 46+). Helps tailor products and marketing messages to specific age demographics.

```python
# Create age groups
df['Age_Group'] = pd.cut(df['Age'], bins=[0, 25, 35, 45, 100],
                        labels=['18-25', '26-35', '36-45', '46+'])

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
sns.barplot(x='Age_Group', y='Purchase_Amount', data=df, hue='Age_Group',
            palette='viridis', errorbar=None, legend=False)
plt.title('Purchase Amount by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Average Purchase Amount ($)')

plt.subplot(2, 2, 2)
sns.barplot(x='Age_Group', y='Frequency_of_Purchase', data=df, hue='Age_Group',
            palette='plasma', errorbar=None, legend=False)
plt.title('Purchase Frequency by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Average Frequency')

plt.subplot(2, 2, 3)
age_income = df.groupby(['Age_Group', 'Income_Level'], observed=False)['Purchase_Amount'].mean().unstack()
age_income.plot(kind='bar', ax=plt.gca(), color=['#1f77b4', '#ff7f0e', '#2ca02c'])
plt.title('Purchase by Age & Income')
plt.xlabel('Age Group')
plt.ylabel('Average Purchase Amount ($)')
plt.legend(title='Income Level', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=0)

plt.subplot(2, 2, 4)
discount_usage = pd.crosstab(df['Age_Group'], df['Discount_Used'], normalize='index')
discount_usage.plot(kind='bar', ax=plt.gca(), color=['#e74c3c', '#27ae60'])
plt.title('Discount Usage by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Proportion')
plt.legend(title='Discount Used', labels=['No', 'Yes'])
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

##  E-Commerce Consumer Behavior - Business Report

## Key Findings

### 1. **Customer Demographics**
- **Age**: Evenly distributed across 18-70 years
- **Gender**: Balanced (approx. **48% Female**, **43% Male**, with smaller non-binary segments)
- **Income**: Split primarily between High and Middle-income levels
- **Education**: Diverse, with High School, Bachelor's, and Master's degrees all well-represented

---

### 2. **Time-Based Patterns**
- **Peak Shopping Days**: **Tuesday** (highest volume ~160), followed by **Sunday** and **Thursday**. Wednesday is the lowest traffic day
- **Highest Revenue Month**: **April** (Top revenue >$28k), followed closely by **August** and **March**. Surprisingly, December is among the lowest revenue months
- **Action**: Schedule major promotions on Tuesdays; prepare inventory for Q2 (April) spikes rather than relying solely on Q4 holidays

---

### 3. **Marketing Channels**
- **Social Media Influence**: **Moderate/Low** impact. Boxplots show identical median spending across High, Medium, and Low influence
- **Ad Engagement**: **Low** impact. Engagement levels do not correlate linearly with higher spending
- **Best Strategy**: The heatmap suggests the "sweet spot" is **Medium Social Influence + Medium Ad Engagement** (Avg Purchase $287). Aggressive, high-frequency marketing does not yield better results than a moderate approach

---

### 4. **Customer Segments (Income × Education)**
Top spending segments from the heatmap analysis:
1. **Middle Income + Bachelor's** = **$287** (Highest Average)
2. **High Income + Bachelor's** = **$281**
3. **High Income + High School** = **$275**

---

### 5. **High-Value Customers (Top 10%)**
- **Demographics**: The "High Value" customer subset is predominantly **Female** (48%) and **High Income**
- **Education Profile**: Interestingly, **High School** graduates make up the largest count of high-value customers, followed by Bachelor's and Master's

---

### 6. **Age Group Behavior**
- **18-25**: Avg spend **~$280** (Highest), **High** discount usage (Yes > No)
- **26-35**: Avg spend **~$278**, **Low** discount usage (No > Yes)
- **36-45**: Avg spend **~$275**, **High** discount usage
- **46+**: Avg spend **~$265** (Lowest), **High** discount usage

---

## Top 3 Recommendations

### 1. **Refine Targeting Strategy**
Shift focus to the **Middle-Income Bachelor's** segment, as they have the highest average order value ($287), rather than assuming High-Income/Master's are the top spenders.

### 2. **Optimize Calendar & Timing**
Move weekly campaign launches to **Tuesday mornings**. Invest heavily in **April** and **August** marketing, as these are organic revenue peaks, and investigate why December performance is lagging.

### 3. **VIP Program**
Create loyalty rewards for top 10% customers
