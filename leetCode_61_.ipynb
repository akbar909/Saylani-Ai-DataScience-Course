### Day 61 : Trapping Rain Water

```python
height = [4,2,0,3,2,5]
```

```python
def trap(height):
    if not height:
        return 0

    l, r = 0, len(height) - 1
    leftMax, rightMax = height[l], height[r]
    res = 0

    while l < r:
        if leftMax < rightMax:
            l += 1
            leftMax = max(leftMax, height[l])
            res += leftMax - height[l]
        else:
            r -= 1
            rightMax = max(rightMax, height[r])
            res += rightMax - height[r]

    return res
```

```python
trap(height)
```

### Day 62 : Wildcard Matching

```python
s = "aa"
p = "a"
```

```python
def isMatch(s, p):
    m, n = len(s), len(p)

    dp = [[False] * (n + 1) for _ in range(m + 1)]

    dp[0][0] = True

    for j in range(1, n + 1):
        if p[j - 1] == '*':
            dp[0][j] = dp[0][j - 1]

    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if p[j - 1] == '*':
                dp[i][j] = dp[i - 1][j] or dp[i][j - 1]
            elif p[j - 1] == '?' or s[i - 1] == p[j - 1]:
                dp[i][j] = dp[i - 1][j - 1]

    return dp[m][n]
```

```python
isMatch(s,p)
```

### Day 63 : Largest Rectangle in Histogram

```python
heights = [2,1,5,6,2,3]
```

```python
def largestRectangleArea(heights):
    maxArea = 0
    stack = []

    for i, h in enumerate(heights):
        start = i
        while stack and stack[-1][1] > h:
            index, height = stack.pop()
            maxArea = max(maxArea, height * (i - index))
            start = index
        stack.append((start, h))

    for i, h in stack:
        maxArea = max(maxArea, h * (len(heights) - i))

    return maxArea
```

```python
largestRectangleArea(heights)
```

### Day 64 : Distinct Subsequences

```python
s = "rabbbit"
t = "rabbit"
```

```python
def numDistinct(s, t):
    cache = {}

    def dfs(i, j):
        if j == len(t):
            return 1
        if i == len(s):
            return 0
        if (i, j) in cache:
            return cache[(i, j)]

        if s[i] == t[j]:
            cache[(i, j)] = dfs(i + 1, j + 1) + dfs(i + 1, j)
        else:
            cache[(i, j)] = dfs(i + 1, j)
        return cache[(i, j)]
    return dfs(0, 0)
```

```python
numDistinct(s,t)
```

### Day 65 : Longest Valid Parentheses

```python
s = ")()())"
```

```python
def longestValidParentheses(s):
    stack = [-1]
    max_length = 0

    for i in range(len(s)):
        if s[i] == '(':
            stack.append(i)
        else:
            stack.pop()
            if not stack:
                stack.append(i)
            else:
                max_length = max(max_length, i - stack[-1])

    return max_length
```

```python
longestValidParentheses(s)
```
