# String Methods Assignment

```python
# capitalize() : Converts the first character to upper case

capt = "ai class"
capt.capitalize()
```

```python
# casefold() : Converts string into lower case

caseFold = "Ai Class"
caseFold.casefold()
```

```python
# center() : Returns a centered string

cent = "Ai-Class"
a= cent.center(20)
print(a)
print("-----------------------")
b= cent.center(20,"O")
print(b)
```

```python
# count() : Returns the number of times a specified value occurs in a string

txt = "I love coding "
x = txt.count("o")
print(x)

print("---------------------")

y = txt.count("o", 6,9)
print(y)
```

```python
# encode : Returns an encoded version of the string

txt = "My name is Akbár"
x = txt.encode()
print(x)

print("----------------------------")

print(txt.encode(encoding="ascii",errors="backslashreplace"))
print(txt.encode(encoding="ascii",errors="ignore"))
print(txt.encode(encoding="ascii",errors="namereplace"))
print(txt.encode(encoding="ascii",errors="replace"))
print(txt.encode(encoding="ascii",errors="xmlcharrefreplace"))
```

```python
# endswith() : 	Returns true if the string ends with the specified value

txt = "Hello, my name is Akbar."
x = txt.endswith(".")
print(x)

print("-------------------------")

y = txt.endswith("my", 7, 9)
print(y)

print("-------------------------")

z = txt.endswith(("name", "Akbar."))
print(z)
```

```python
# expandtabs() :	Sets the tab size of the string

txt = "A\tK\tB\tA\tR"
x =  txt.expandtabs()
print(x)

print("------------------------------")

y =  txt.expandtabs(3)
print(y)

print("------------------------------")

print(txt)
print("------------------------------")
print(txt.expandtabs())
print("------------------------------")
print(txt.expandtabs(2))
print("------------------------------")
print(txt.expandtabs(4))
print("------------------------------")
print(txt.expandtabs(10))
```

```python
# find() :	Searches the string for a specified value and returns the position of where it was found

txt = "Hello, my name is Akbar."
x = txt.find("name")
print(x)

print("----------------------")
y = txt.find("e", 2, 10)
print(y)
```

```python
# format() :	Formats specified values in a string 

txt = "My age is {age}!"
print(txt.format(age = 20))

txt1 = "My name is {fname}, I'm {age}".format(fname = "Akbar", age = 20)
print(txt1)
txt2 = "My name is {0}, I'm {1}".format("Akbar",20)
print(txt2)
txt3 = "My name is {}, I'm {}".format("Akbar",20)
print(txt3)
```

```python
# format_map() :	Formats specified values in a string
data = {'name': 'Ghulam Akbar', 'age': 20}
message = "Hello, my name is {name} and I am {age} years old."
formatted_message = message.format_map(data)
print(formatted_message)
```

```python
# index() :	Searches the string for a specified value and returns the position of where it was found

txt = "Hello, my name is Akbar."
x = txt.index("Akbar")
print(x)

y = txt.index("e")
print(y)

z = txt.index("A", 15, 20)
print(z)
```

```python
# isalnum() : 	Returns True if all characters in the string are alphanumeric

txt = "AiClass12"
txt1 = "AiClass 12"
x = txt.isalnum()
y = txt1.isalnum()
print(x)
print(y)
```

```python
# isalpha() :	Returns True if all characters in the string are in the alphabet

txt = "AiclassJ"
txt1 = "AiClass10"

x = txt.isalpha()
print(x)
y = txt1.isalpha()
print(y)
```

```python
# isascii() :	Returns True if all characters in the string are ascii characters

txt = "AiClass123"
x = txt.isascii()
print(x)
```

```python
# isdecimal():	Returns True if all characters in the string are decimals

txt = "1234"
x = txt.isdecimal()
print(x)

a = "\u0030" #unicode for 0
b = "\u0047" #unicode for G

print(a.isdecimal())
print(b.isdecimal())
```

```python
# isdigit() :	Returns True if all characters in the string are digits

txt = "50800"
x = txt.isdigit()
print(x)

a = "\u0030" #unicode for 0
b = "\u00B2" #unicode for ²

print(a.isdigit())
print(b.isdigit())
```

```python
# isidentifier() :	Returns True if the string is an identifier

txt = "Demo"
x = txt.isidentifier()
print(x)

a = "MyFolder"
b = "Demo002"
c = "2bring"
d = "my demo"

print(a.isidentifier())
print(b.isidentifier())
print(c.isidentifier())
print(d.isidentifier())
```

```python
# islower() : Returns True if all characters in the string are lower case

txt = "akbar"
x = txt.islower()
print(x)

a = "Ghulam Akbar!"
b = "hello 123"
c = "mynameisAkbar"

print(a.islower())
print(b.islower())
print(c.islower())
```

```python
# isnumeric() : 	Returns True if all characters in the string are numeric

txt = "565543"
x = txt.isnumeric()
print(x)

a = "\u0030" #unicode for 0
b = "\u00B2" #unicode for &sup2;
c = "10km2"
d = "-1"
e = "1.5"

print(a.isnumeric())
print(b.isnumeric())
print(c.isnumeric())
print(d.isnumeric())
print(e.isnumeric())
```

```python
# isprintable() :	Returns True if all characters in the string are printable

txt = "Hello! Are you #1?"
txt1 = "Hello!\nAre you #1?"

x = txt.isprintable()
print(x)

y = txt1.isprintable()
print(y)
```

```python
# isspace() :	Returns True if all characters in the string are whitespaces

txt = "   "
txt1 = "   s   "

x = txt.isspace()
print(x)

y = txt1.isspace()
print(y)
```

```python
# istitle() :	Returns True if the string follows the rules of a title

txt = "Hello, And Welcome To My World!"
x = txt.istitle()
print(x)

a = "HELLO, AND WELCOME TO MY WORLD"
b = "Hello"
c = "22 Names"
d = "This Is %'!?"

print(a.istitle())
print(b.istitle())
print(c.istitle())
print(d.istitle())
```

```python
# isupper()	Returns True if all characters in the string are upper case

txt = "AKBAR"
x = txt.isupper()
print(x)

a = "Hello World!"
b = "hello 123"
c = "MY NAME IS PETER"

print(a.isupper())
print(b.isupper())
print(c.isupper())
```

```python
# join() :	Joins the elements of an iterable to the end of the string

myTuple = ("Ahmed", "Ali", "Akbar")

x = " ".join(myTuple)

print(x)

myDict = {"name": "John", "country": "Norway"}
mySeparator = "TEST"

x = mySeparator.join(myDict)

print(x)
```

```python
# ljust() :	Returns a left justified version of the string
txt = "banana"

x = txt.ljust(20)
y = txt.ljust(20, "1")

print(x, "is my favorite fruit.")
print(y, "is my favorite fruit.")
```

```python
# lower() :  Converts a string into lower case

txt = "Hello my FRIENDS"

x = txt.lower()

print(x)
```

```python
# lstrip() :	Returns a left trim version of the string

txt = "     banana     "

x = txt.lstrip()

print("of all fruits", x, "is my favorite")

txt = ",,,,,ssaaww.....banana"

x = txt.lstrip(",.asw")

print(x)
```

```python
# maketrans() : 	Returns a translation table to be used in translations
txt = "Hello Sam!"
mytable = str.maketrans("S", "P")
print(txt.translate(mytable))

txt = "Hi Sam!"
x = "mSa"
y = "eJo"
mytable = str.maketrans(x, y)
print(txt.translate(mytable))

txt = "Good night Sam!"
x = "mSa"
y = "eJo"
z = "odnght"
mytable = str.maketrans(x, y, z)
print(txt.translate(mytable))

txt = "Good night Sam!"
x = "mSa"
y = "eJo"
z = "odnght"
print(str.maketrans(x, y, z))
```

```python
# partition() :	Returns a tuple where the string is parted into three parts

txt = "I could eat bananas all day"
x = txt.partition("eat")
print(x)

txt = "I could eat bananas all day"
x = txt.partition("apples")
print(x)
```

```python
# replace() :	Returns a string where a specified value is replaced with a specified value

txt = "I like Gaming"
x = txt.replace("Gaming", "Coding")
print(x)

txt = "one one was a race horse, two two was one too."
x = txt.replace("one", "three", 2)
print(x)
```

```python
# rfind() :	Searches the string for a specified value and returns the last position of where it was found

txt = "Mi Akbar, bye Akbar."
x = txt.rfind("Akbar")
print(x)

txt = "Hello, welcome to my world."
x = txt.rfind("e", 5, 10)
print(x)

txt = "Hello, welcome to my world."

print(txt.rfind("q"))
# print(txt.rindex("q")) # this will raise an exception
```

```python
# rindex() :	Searches the string for a specified value and returns the last position of where it was found

txt = "Mi casa, su casa."
x = txt.rindex("casa")
print(x)
```

```python
# rjust() :	Returns a right justified version of the string

txt = "banana"
x = txt.rjust(20)
print(x, "is my favorite fruit.")

txt = "banana"
x = txt.rjust(20, "O")
print(x)
```

```python
# rpartition() :	Returns a tuple where the string is parted into three parts

txt = "I could eat bananas all day, bananas are my favorite fruit"
x = txt.rpartition("bananas")
print(x)

txt = "I could eat bananas all day, bananas are my favorite fruit"
x = txt.rpartition("apples")
print(x)
```

```python
# rsplit() :	Splits the string at the specified separator, and returns a list

txt = "apple, banana, cherry"
x = txt.rsplit(", ")
print(x)

txt = "apple, banana, cherry"
x = txt.rsplit(", ", 1)
print(x)
```

```python
# rstrip() :	Returns a right trim version of the string

txt = "     banana     "
x = txt.rstrip()
print("of all fruits", x, "is my favorite")

txt = "..banana,,,,,ssqqqww....."
x = txt.rstrip(",.qsw")
print(x)
```

```python
# split() : 	Splits the string at the specified separator, and returns a list

txt = "welcome to the jungle"
x = txt.split()
print(x)

txt = "hello, my name is Peter, I am 26 years old"
x = txt.split(", ")
print(x)

txt = "apple#banana#cherry#orange"
x = txt.split("#")
print(x)
```

```python
# splitlines() :	Splits the string at line breaks and returns a list

txt = "Thank you for the music\nWelcome to the jungle"
x = txt.splitlines()
print(x)

txt = "Thank you for the music\nWelcome to the jungle"
# Split the string, but keep the line breaks
x = txt.splitlines(True)
print(x)
```

```python
# startswith() :	Returns true if the string starts with the specified value


txt = "Hello, welcome to my world."
x = txt.startswith("Hello")
print(x)

txt = "Hello, welcome to my world."
x = txt.startswith("wel", 7, 20)
print(x)

txt = "Hi, welcome to my world."
x = txt.startswith(("Hello", "Hi"))
print(x)
```

```python
# strip() : 	Returns a trimmed version of the string

txt = "     banana     "
x = txt.strip()
print("of all fruits", x, "is my favorite")

txt = ",,,,,rrttgg.....banana....rrr"
x = txt.strip(",.grt")
print(x)
```

```python
# swapcase() :	Swaps cases, lower case becomes upper case and vice versa

txt = "Hello My Name Is PETER"
x = txt.swapcase()
print(x)
```

```python
# title():	Converts the first character of each word to upper case

txt = "Welcome to my world"
x = txt.title()
print(x)

txt = "Welcome to my 2nd world"
x = txt.title()
print(x)

txt = "hello b2b2b2 and 3g3g3g"
x = txt.title()
print(x)
```

```python
# translate():	Returns a translated string

txt = "Hello Sam!"
mytable = str.maketrans("S", "P")
print(txt.translate(mytable))
#use a dictionary with ascii codes to replace 83 (S) with 80 (P):
mydict = {83:  80}
txt = "Hello Sam!"
print(txt.translate(mydict))

txt = "Hi Sam!"
x = "mSa"
y = "eJo"
mytable = str.maketrans(x, y)
print(txt.translate(mytable))

txt = "Good night Sam!"
x = "mSa"
y = "eJo"
z = "odnght"
mytable = str.maketrans(x, y, z)
print(txt.translate(mytable))
```

```python
# upper():	Converts a string into upper case

txt = "Hello my friends"
x = txt.upper()
print(x)
```

```python
# zfill(): Fills the string with a specified number of 0 values at the beginning

txt = "50"
# Fill the string with zeros until it is 10 characters long
x = txt.zfill(10)
print(x)

a = "hello"
b = "welcome to the jungle"
c = "10.000"

print(a.zfill(10))
print(b.zfill(10))
print(c.zfill(10))
```
